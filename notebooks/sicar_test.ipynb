```python
import pandas as pd
import geopandas as gpd
from shapely import wkt
from hooks.http_sicar_hook import HttpSicarHook
```

```python
sicar = HttpSicarHook()
```

```python
data = sicar.get_sicar_data(
    uf="AC",
    # is_pagination=False
    is_pagination=True
)
```

```python
df = gpd.GeoDataFrame(
    data,
    geometry=[wkt.loads(g) if g else None for g in (r["geometry"] for r in data)],
    crs="EPSG:4326",
)
```

```python
df.to_pickle("output/treinamento_modelo_sicar_ac.pkl")
df.to_csv("output/treinamento_modelo_sicar_ac.txt", index=False, sep=";")
df.to_parquet("output/treinamento_modelo_sicar_ac.parquet", index=False)
```
